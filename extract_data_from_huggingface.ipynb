```python
from datasets import load_dataset
import pandas as pd

# Load dataset from Hugging Face
dataset = load_dataset("lukebarousse/data_jobs")

# Check what splits are available
print(dataset)

# Convert one split (e.g., 'train') to pandas DataFrame
df = dataset['train'].to_pandas()

# Display info
print(df.info())
print(df.head())
```

```python
df.to_csv("data_jobs.csv", index=False)
print("✅ Dataset saved as data_jobs.csv")
```
