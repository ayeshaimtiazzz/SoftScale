```python
import os
import pandas as pd

# Path to your folder
folder_path = "raw_data_talent_match"

# Create an empty list to store dataframes
all_data = []

# Loop through files
for file in os.listdir(folder_path):
    file_path = os.path.join(folder_path, file)

    try:
        # Read file depending on extension
        if file.endswith(".csv"):
            df = pd.read_csv(file_path, encoding="utf-8", on_bad_lines="skip")
        elif file.endswith((".xls", ".xlsx")):
            df = pd.read_excel(file_path)
        else:
            print(f"[INFO] Skipping unsupported file: {file}")
            continue

        print(f"[INFO] Processing file: {file}")
        df.columns = [col.lower().strip() for col in df.columns]  # Normalize column names

        # Identify job title and description columns
        title_col = None
        desc_col = None

        for col in df.columns:
            if "job_title" in col or "title" in col or "Job Title" in col or "jobtitle" in col:
                title_col = col
            if "description" in col or "job_description" in col or "Description" in col or "Job Description" in col or "jobdescription" in col:
                desc_col = col

        if title_col and desc_col:
            temp_df = df[[title_col, desc_col]].copy()
            temp_df.rename(columns={title_col: "job_title", desc_col: "job_description"}, inplace=True)
            all_data.append(temp_df)
        else:
            print(f"[WARNING] Skipping {file} (required columns not found).")

    except Exception as e:
        print(f"[ERROR] Could not read {file}: {e}")

# Combine all extracted dataframes
if all_data:
    combined_df = pd.concat(all_data, ignore_index=True)
    combined_df.dropna(subset=["job_title", "job_description"], inplace=True)
    print(f"\n[INFO] Combined dataset created with {len(combined_df)} entries.")

    # Save to CSV
    combined_df.to_csv("combined_jobs.csv", index=False, encoding="utf-8")
    print("[INFO] Combined dataset saved as 'combined_jobs.csv'")
else:
    print("[ERROR] No valid data found to combine.")
```
