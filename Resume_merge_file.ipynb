```python
def extract_skills(text):
    if not isinstance(text, str):
        return None
    skills_list = ['Python', 'Java', 'C++', 'SQL', 'Excel', 'Communication', 'Sales', 'Data Analysis',
                   'Machine Learning', 'Project Management', 'Networking', 'Electrical', 'Construction',
                   'Accounting', 'Finance', 'Web Development', 'JavaScript', 'React', 'Node.js', 'UX', 'UI']
    skills_found = [skill for skill in skills_list if skill.lower() in text.lower()]
    return ', '.join(skills_found)

def extract_experience(text):
    if not isinstance(text, str):
        return None
    exp = re.findall(r'(\d+)\s*(?:years|yrs|year)', text, re.IGNORECASE)
    return exp[0] if exp else None

def extract_education(text):
    if not isinstance(text, str):
        return None
    degrees = ['Bachelors', 'Bachelor', 'B.Sc', 'M.Sc', 'Masters', 'MBA', 'PhD', 'High School', 'Diploma']
    edu_found = [degree for degree in degrees if degree.lower() in text.lower()]
    return ', '.join(edu_found)

def extract_age(text):
    if not isinstance(text, str):
        return None
    age_match = re.findall(r'(\d{2})\s*years?', text)
    return age_match[0] if age_match else None

def extract_country(text):
    if not isinstance(text, str):
        return None
    countries = ['USA', 'United States', 'UK', 'Canada', 'India', 'Germany', 'France', 'Australia']
    for country in countries:
        if country.lower() in text.lower():
            return country
    return None

def extract_domain(text):
    if not isinstance(text, str):
        return None
    domains = ['Sales', 'Engineering', 'Construction', 'Data', 'IT', 'Healthcare', 'Finance', 'Marketing',
               'Accounting', 'Web Development', 'Software', 'UX']
    for domain in domains:
        if domain.lower() in text.lower():
            return domain
    return None

# -----------------------------
# Apply Extraction to df3
# -----------------------------
```

```python
df2['skills'] = df2['resume_text'].apply(extract_skills)
df2['experience'] = df2['resume_text'].apply(extract_experience)
df2['education'] = df2['resume_text'].apply(extract_education)
df2['age'] = df2['resume_text'].apply(extract_age)
df2['country'] = df2['resume_text'].apply(extract_country)
df2['domain'] = df2['resume_text'].apply(extract_domain)
```
